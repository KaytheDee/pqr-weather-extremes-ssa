### Rename Columns  

```python
def rename_weather_df(df, location_name):
    df = df.copy()
    rename_dict = {}

    # Dynamically rename columns containing 'Interpol_Precip_' and 'Interpol_Temp_'
    for col in df.columns:
        if 'Interpol_Precip_' in col:
            rename_dict[col] = 'Interpol_Precip'
        if 'Interpol_Temp_' in col:
            rename_dict[col] = 'Interpol_Temp'

    df.rename(columns=rename_dict, inplace=True)
    df['Location'] = location_name
    return df
```

## Variable comparison of weather stations in close proximity  

```python

def resample_and_plot_temp_multi(df1, df2=None, df3=None, freq='D', title='Temperature Over Time', labels=None):

    def prepare(df):
        df = df.copy()
        df['Hour'] = pd.to_datetime(df['Hour'])
        df.set_index('Hour', inplace=True)
        return df['Interpol_Temp'].resample(freq).mean()

    # Prepare and collect all valid DataFrames
    series_list = [prepare(df1)]
    if df2 is not None:
        series_list.append(prepare(df2))
    if df3 is not None:
        series_list.append(prepare(df3))

    if labels is None:
        labels = [f"Dataset {i+1}" for i in range(len(series_list))]

    # Plot
    plt.figure(figsize=(14, 6))
    for series, label in zip(series_list, labels):
        series.plot(label=label)

    plt.title(title)
    plt.xlabel('Date' if freq == 'D' else 'Month')
    plt.ylabel('Average Temperature (°C)', labelpad=10)
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.show()

    return
```

```python
def resample_and_plot_precip_multi(df1, df2=None, df3=None, freq='D', title='Precipitation Over Time', labels=None):

    def prepare(df):
        df = df.copy()
        df['Hour'] = pd.to_datetime(df['Hour'])
        df.set_index('Hour', inplace=True)
        return df['Interpol_Precip'].resample(freq).sum()

    # Prepare and collect all valid DataFrames
    series_list = [prepare(df1)]
    if df2 is not None:
        series_list.append(prepare(df2))
    if df3 is not None:
        series_list.append(prepare(df3))

    if labels is None:
        labels = [f"Dataset {i+1}" for i in range(len(series_list))]

    # Plot
    plt.figure(figsize=(14, 6))
    for series, label in zip(series_list, labels):
        series.plot(label=label)

    plt.title(title)
    plt.xlabel('Date' if freq == 'D' else 'Month')
    plt.ylabel('Total Precipitation', labelpad = 10)
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.show()

    return 
```

```python
def print_precip_stats(df, column='Interpol_Precip'):
    mean_val = df[column].mean()
    std_val = df[column].std()
    print(f"Mean of {column}: {mean_val:.2f}")
    print(f"Standard Deviation of {column}: {std_val:.2f}")
```

```python
def calc_precip_corr(df1, df2, column='Interpol_Precip'):
    common_index = df1.index.intersection(df2.index)
    series1 = df1.loc[common_index, column]
    series2 = df2.loc[common_index, column]
    corr = series1.corr(series2)
    print(f"Correlation between {column} in the two DataFrames: {corr:.3f}")
    return 
```

```python
from scipy.stats import ttest_ind

def run_ttest(df1, df2, column='Interpol_Precip', label1='DF1', label2='DF2'):

    # Drop NaNs to ensure clean comparison
    x = df1[column].dropna()
    y = df2[column].dropna()

    # Run Welch’s t-test (doesn't assume equal variance)
    t_stat, p_val = ttest_ind(x, y, equal_var=False)

    print(f"t-test comparing {label1} and {label2} on '{column}':")
    print(f"  t-statistic: {t_stat:.3f}")
    print(f"  p-value: {p_val:.3f}")

    if p_val < 0.05:
        print("  ✅ Statistically significant difference in means.")
    else:
        print("  ❌ No statistically significant difference in means.")

    return 
```

```python
def calc_precip_corr_three(df1, df2, df3, column='Interpol_Precip'):
    # Find common index across all three
    common_index = df1.index.intersection(df2.index).intersection(df3.index)

    # Extract aligned series
    s1 = df1.loc[common_index, column]
    s2 = df2.loc[common_index, column]
    s3 = df3.loc[common_index, column]

    # Compute pairwise correlations
    corr_12 = s1.corr(s2)
    corr_13 = s1.corr(s3)
    corr_23 = s2.corr(s3)

    # Print results
    print(f"Correlation between df1 and df2: {corr_12:.3f}")
    print(f"Correlation between df1 and df3: {corr_13:.3f}")
    print(f"Correlation between df2 and df3: {corr_23:.3f}")

    return 
```
